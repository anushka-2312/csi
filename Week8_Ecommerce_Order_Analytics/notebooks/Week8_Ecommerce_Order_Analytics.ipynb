```python
## 1.Customers Dataset
# Number of customers
num_customers = 500

customer_types = ["REGULAR", "PREMIUM", "VIP"]

first_names = [
    "Aarav", "Vivaan", "Aditya", "Arjun", "Rahul",
    "Rohan", "Karan", "Ananya", "Priya", "Sneha",
    "Isha", "Neha", "Pooja", "Aditi", "Kavya"
]

last_names = [
    "Sharma", "Patil", "Deshmukh", "Joshi", "Kulkarni",
    "Mehta", "Shah", "Verma", "Gupta", "Singh"
]

customers = []

for i in range(1, num_customers + 1):

    customer_id = f"C{i:04d}"

    name = random.choice(first_names) + " " + random.choice(last_names)

    email = (
        name.lower()
        .replace(" ", ".")
        + str(random.randint(1, 999))
        + "@gmail.com"
    )

    # Approximately 2% invalid emails
    if i % 50 == 0:
        email = email.replace("@", "")

    registration_date = (
        datetime.now() - timedelta(days=random.randint(1, 1000))
    ).strftime("%Y-%m-%d")

    customer_type = random.choice(customer_types)

    customers.append([
        customer_id,
        name,
        email,
        registration_date,
        customer_type
    ])

customers_df = pd.DataFrame(
    customers,
    columns=[
        "customer_id",
        "customer_name",
        "email",
        "registration_date",
        "customer_type"
    ]
)

customers_df.to_csv(
    "data/customers.csv",
    index=False
)

print("Customers generated:", len(customers_df))
customers_df.head()
```

```python
## 2.Products Dataset
num_products = 500

categories = {
    "Electronics": ["Laptop", "Mobile", "Headphones", "Camera", "Keyboard"],
    "Clothing": ["T-Shirt", "Jeans", "Jacket", "Shoes", "Dress"],
    "Home": ["Chair", "Table", "Lamp", "Curtains", "Cushion"],
    "Books": ["Novel", "Textbook", "Biography", "Comics", "Dictionary"]
}

products = []

for i in range(1, num_products + 1):

    product_id = f"P{i:04d}"

    category = random.choice(list(categories.keys()))
    product_name = random.choice(categories[category])

    # Add intentional data-quality problems
    if i % 20 == 0:
        product_name = "  " + product_name.lower() + "  "
    elif i % 25 == 0:
        product_name = product_name.upper()

    subcategory = category + " Products"

    cost_price = round(random.uniform(100, 5000), 2)

    products.append([
        product_id,
        product_name,
        category,
        subcategory,
        cost_price
    ])

products_df = pd.DataFrame(
    products,
    columns=[
        "product_id",
        "product_name",
        "category",
        "subcategory",
        "cost_price"
    ]
)

products_df.to_csv(
    "data/products.csv",
    index=False
)

print("Products generated:", len(products_df))
products_df.head()
```

```python
# 3.Orders Dataset
num_orders = 1000

statuses = [
    "PLACED",
    "SHIPPED",
    "DELIVERED",
    "CANCELLED",
    "RETURNED"
]

regions = ["NORTH", "SOUTH", "EAST", "WEST"]

order_ids = []
orders = []

for i in range(1, num_orders + 1):

    order_id = f"O{i:05d}"
    order_ids.append(order_id)

    # 5% missing customer IDs
    if i % 20 == 0:
        customer_id = None
    else:
        customer_id = random.choice(customers_df["customer_id"].tolist())

    order_date = datetime.now() - timedelta(
        days=random.randint(1, 450),
        hours=random.randint(0, 23),
        minutes=random.randint(0, 59)
    )

    order_date_str = order_date.strftime("%Y-%m-%d %H:%M:%S")

    # Some dates intentionally in DD-MM-YYYY format
    if i % 100 == 0:
        order_date_str = order_date.strftime("%d-%m-%Y")

    status = random.choice(statuses)
    region = random.choice(regions)

    orders.append([
        order_id,
        customer_id,
        order_date_str,
        status,
        region
    ])

orders_df = pd.DataFrame(
    orders,
    columns=[
        "order_id",
        "customer_id",
        "order_date",
        "status",
        "region_code"
    ]
)

orders_df.to_csv(
    "data/orders.csv",
    index=False
)

print("Orders generated:", len(orders_df))
orders_df.head()
```

```python
# 4.Order Items Dataset
num_items = 2000

product_ids = products_df["product_id"].tolist()

order_items = []

for i in range(1, num_items + 1):

    item_id = f"I{i:05d}"

    # Select an existing order
    order_id = random.choice(order_ids)

    product_id = random.choice(product_ids)

    quantity = random.randint(1, 5)

    # Approximately 3% negative quantities
    if i % 33 == 0:
        quantity = -quantity

    unit_price = round(random.uniform(200, 10000), 2)

    discount_percent = round(
        random.uniform(0, 100), 2
    )

    order_items.append([
        item_id,
        order_id,
        product_id,
        quantity,
        unit_price,
        discount_percent
    ])

order_items_df = pd.DataFrame(
    order_items,
    columns=[
        "item_id",
        "order_id",
        "product_id",
        "quantity",
        "unit_price",
        "discount_percent"
    ]
)

order_items_df.to_csv(
    "data/order_items.csv",
    index=False
)

print("Order items generated:", len(order_items_df))
order_items_df.head()
```

```python
def clean_orders(df):
    """
    Cleans the orders dataset.

    Operations:
    1. Convert all order dates to YYYY-MM-DD HH:MM:SS
    2. Preserve missing customer IDs as NULL
    3. Remove duplicate orders
    """
    
    cleaned = df.copy()

    # Count missing customer IDs before cleaning
    missing_customer_count = cleaned["customer_id"].isna().sum()

    # Convert dates.
    # dayfirst=True allows DD-MM-YYYY values to be interpreted correctly.
    cleaned["order_date"] = pd.to_datetime(
        cleaned["order_date"],
        dayfirst=True,
        errors="coerce"
    )

    # Convert dates to standard format
    cleaned["order_date"] = cleaned["order_date"].dt.strftime(
        "%Y-%m-%d %H:%M:%S"
    )

    # Keep missing customer IDs as NULL
    cleaned["customer_id"] = cleaned["customer_id"].replace(
        ["", "NULL", "null", "None"],
        np.nan
    )

    # Remove duplicate order IDs
    duplicate_orders = cleaned["order_id"].duplicated().sum()

    cleaned = cleaned.drop_duplicates(
        subset=["order_id"]
    )

    return cleaned, missing_customer_count, duplicate_orders
```

```python
orders_cleaned, missing_customers, duplicate_orders = clean_orders(
    orders_df
)

print("Orders cleaned successfully!")
print("Missing customer IDs:", missing_customers)
print("Duplicate orders removed:", duplicate_orders)

orders_cleaned.head()
```

```python
## 8. Clean Products Dataset
def clean_products(df):
    """
    Cleans product names by:
    - Removing leading/trailing spaces
    - Converting names to title case
    """
    
    cleaned = df.copy()

    # Count product names that need cleaning
    original_names = cleaned["product_name"].copy()

    # Remove extra spaces
    cleaned["product_name"] = (
        cleaned["product_name"]
        .astype(str)
        .str.strip()
    )

    # Normalize capitalization
    cleaned["product_name"] = (
        cleaned["product_name"]
        .str.title()
    )

    changed_names = (
        original_names != cleaned["product_name"]
    ).sum()

    return cleaned, changed_names
```

```python
products_cleaned, changed_product_names = clean_products(
    products_df
)

print("Products cleaned successfully!")
print("Product names normalized:", changed_product_names)

products_cleaned.head(15)
```

```python
## 9. Validate Customer Emails
def validate_emails(df):
    """
    Returns customer IDs with invalid email addresses.
    """
    
    email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

    valid_email = df["email"].astype(str).str.match(
        email_pattern,
        na=False
    )

    invalid_customer_ids = df.loc[
        ~valid_email,
        "customer_id"
    ].tolist()

    return invalid_customer_ids
```

```python
invalid_customer_ids = validate_emails(
    customers_df
)

print("Invalid email count:", len(invalid_customer_ids))

print("\nInvalid customer IDs:")
print(invalid_customer_ids)
```

```python
## 10. Check Referential Integrity
def check_referential_integrity(
    orders_df,
    order_items_df
):
    """
    Finds order items whose order_id
    does not exist in the orders dataset.
    """

    valid_order_ids = set(
        orders_df["order_id"]
    )

    invalid_items = order_items_df[
        ~order_items_df["order_id"].isin(
            valid_order_ids
        )
    ]

    return invalid_items
```

```python
invalid_order_items = check_referential_integrity(
    orders_cleaned,
    order_items_df
)

print(
    "Referential integrity issues:",
    len(invalid_order_items)
)

invalid_order_items.head()
```

```python
## 11. Clean Order Items Dataset
def clean_order_items(df):
    cleaned = df.copy()

    # Convert numeric columns
    cleaned["quantity"] = pd.to_numeric(
        cleaned["quantity"],
        errors="coerce"
    )

    cleaned["unit_price"] = pd.to_numeric(
        cleaned["unit_price"],
        errors="coerce"
    )

    cleaned["discount_percent"] = pd.to_numeric(
        cleaned["discount_percent"],
        errors="coerce"
    )

    # Remove exact duplicate rows
    cleaned = cleaned.drop_duplicates()

    return cleaned
```

```python
order_items_cleaned = clean_order_items(
    order_items_df
)

print("Order items cleaned successfully!")
print(order_items_cleaned.shape)

order_items_cleaned.head()
```

```python
query = """
SELECT
    p.product_id,
    p.product_name,

    SUM(
        CASE
            WHEN oi.quantity > 0
            THEN oi.quantity
            ELSE 0
        END
    ) AS purchase_quantity,

    ABS(
        SUM(
            CASE
                WHEN oi.quantity < 0
                THEN oi.quantity
                ELSE 0
            END
        )
    ) AS return_quantity

FROM products p
JOIN order_items oi
    ON p.product_id = oi.product_id

GROUP BY
    p.product_id,
    p.product_name

HAVING return_quantity > purchase_quantity

ORDER BY return_quantity DESC;
"""

products_more_returns = pd.read_sql_query(
    query,
    connection
)

products_more_returns
```

```python
query = """
SELECT
    p.category,

    ABS(
        SUM(
            CASE
                WHEN oi.quantity < 0
                THEN oi.quantity
                ELSE 0
            END
        )
    ) AS returned_items,

    SUM(
        ABS(oi.quantity)
    ) AS total_items,

    ROUND(
        100.0 *
        ABS(
            SUM(
                CASE
                    WHEN oi.quantity < 0
                    THEN oi.quantity
                    ELSE 0
                END
            )
        )
        /
        NULLIF(
            SUM(ABS(oi.quantity)),
            0
        ),
        2
    ) AS return_rate_percent

FROM products p

JOIN order_items oi
    ON p.product_id = oi.product_id

GROUP BY p.category

ORDER BY return_rate_percent DESC;
"""

return_rate_by_category = pd.read_sql_query(
    query,
    connection
)

return_rate_by_category
```

```python
# Running Totals with Window Functions
# Running Total of Revenue by Month
query = """
WITH monthly_revenue AS (
    SELECT
        strftime('%Y-%m', o.order_date) AS order_month,

        SUM(
            oi.quantity *
            oi.unit_price *
            (1 - oi.discount_percent / 100.0)
        ) AS monthly_revenue

    FROM orders o

    JOIN order_items oi
        ON o.order_id = oi.order_id

    GROUP BY order_month
)

SELECT
    order_month,

    ROUND(monthly_revenue, 2)
        AS monthly_revenue,

    ROUND(
        SUM(monthly_revenue) OVER (
            ORDER BY order_month
            ROWS BETWEEN UNBOUNDED PRECEDING
            AND CURRENT ROW
        ),
        2
    ) AS running_total_revenue

FROM monthly_revenue

ORDER BY order_month;
"""

running_revenue = pd.read_sql_query(
    query,
    connection
)

running_revenue
```

```python
#8. Ranking with DENSE_RANK
# DENSE_RANK for Top Products
query = """
WITH product_revenue AS (

    SELECT
        p.product_id,
        p.product_name,
        p.category,

        SUM(
            oi.quantity *
            oi.unit_price *
            (1 - oi.discount_percent / 100.0)
        ) AS total_revenue

    FROM products p

    JOIN order_items oi
        ON p.product_id = oi.product_id

    GROUP BY
        p.product_id,
        p.product_name,
        p.category
),

ranked_products AS (

    SELECT
        product_id,
        product_name,
        category,
        ROUND(total_revenue, 2) AS total_revenue,

        DENSE_RANK() OVER (
            PARTITION BY category
            ORDER BY total_revenue DESC
        ) AS revenue_rank

    FROM product_revenue
)

SELECT *
FROM ranked_products

WHERE revenue_rank <= 3

ORDER BY
    category,
    revenue_rank;
"""

top_products_by_category = pd.read_sql_query(
    query,
    connection
)

top_products_by_category
```

```python
# 9 LAG and LEAD
query = """
WITH monthly_revenue AS (

    SELECT
        strftime('%Y-%m', o.order_date) AS order_month,

        SUM(
            oi.quantity *
            oi.unit_price *
            (1 - oi.discount_percent / 100.0)
        ) AS revenue

    FROM orders o

    JOIN order_items oi
        ON o.order_id = oi.order_id

    GROUP BY order_month
)

SELECT
    order_month,

    ROUND(revenue, 2) AS revenue,

    ROUND(
        LAG(revenue) OVER (
            ORDER BY order_month
        ),
        2
    ) AS previous_month_revenue,

    ROUND(
        LEAD(revenue) OVER (
            ORDER BY order_month
        ),
        2
    ) AS next_month_revenue

FROM monthly_revenue

ORDER BY order_month;
"""

monthly_revenue_comparison = pd.read_sql_query(
    query,
    connection
)

monthly_revenue_comparison
```

```python
#Query 15 — Cohort Analysis
query = """
WITH customer_first_purchase AS (

    SELECT
        customer_id,
        MIN(date(order_date)) AS first_purchase_date

    FROM orders

    WHERE customer_id IS NOT NULL

    GROUP BY customer_id
),

customer_cohorts AS (

    SELECT
        customer_id,

        strftime(
            '%Y-%m',
            first_purchase_date
        ) AS cohort_month

    FROM customer_first_purchase
),

customer_activity AS (

    SELECT DISTINCT
        o.customer_id,

        strftime(
            '%Y-%m',
            o.order_date
        ) AS activity_month

    FROM orders o

    WHERE o.customer_id IS NOT NULL
),

cohort_activity AS (

    SELECT
        cc.cohort_month,
        ca.activity_month,
        COUNT(DISTINCT ca.customer_id)
            AS active_customers

    FROM customer_cohorts cc

    JOIN customer_activity ca
        ON cc.customer_id = ca.customer_id

    GROUP BY
        cc.cohort_month,
        ca.activity_month
)

SELECT
    cohort_month,
    activity_month,
    active_customers

FROM cohort_activity

ORDER BY
    cohort_month,
    activity_month;
"""

cohort_analysis = pd.read_sql_query(
    query,
    connection
)

cohort_analysis.head(30)
```

```python
def generate_report(report_type, start_date, end_date):

    if report_type.lower() not in ["daily", "weekly", "monthly"]:
        print("Invalid report type.")
        print("Choose: daily, weekly, or monthly")
        return

    query = """
    SELECT
        COUNT(DISTINCT o.order_id) AS total_orders,

        ROUND(
            SUM(
                oi.quantity *
                oi.unit_price *
                (1 - oi.discount_percent / 100.0)
            ),
            2
        ) AS total_revenue,

        SUM(
            CASE
                WHEN oi.quantity < 0
                THEN ABS(oi.quantity)
                ELSE 0
            END
        ) AS returned_items

    FROM orders o

    JOIN order_items oi
        ON o.order_id = oi.order_id

    WHERE date(o.order_date)
          BETWEEN date(?) AND date(?);
    """

    result = pd.read_sql_query(
        query,
        connection,
        params=[start_date, end_date]
    )

    print("\n================================")
    print("       E-COMMERCE REPORT")
    print("================================")
    print("Report Type :", report_type.upper())
    print("Start Date  :", start_date)
    print("End Date    :", end_date)
    print("--------------------------------")

    print(
        "Total Orders :",
        result["total_orders"].iloc[0]
    )

    print(
        "Total Revenue:",
        result["total_revenue"].iloc[0]
    )

    print(
        "Returned Items:",
        result["returned_items"].iloc[0]
    )

    print("================================\n")

    return result
```

```python
import sqlite3
from datetime import datetime
```

```python
def test_invalid_order_reference():
    conn = sqlite3.connect(":memory:")
    cursor = conn.cursor()

    # Create tables
    cursor.execute("""
        CREATE TABLE orders (
            order_id TEXT PRIMARY KEY
        )
    """)

    cursor.execute("""
        CREATE TABLE order_items (
            order_id TEXT,
            product_id TEXT,
            quantity INTEGER
        )
    """)

    # Valid orders
    cursor.executemany(
        "INSERT INTO orders VALUES (?)",
        [
            ("ORD001",),
            ("ORD002",)
        ]
    )

    # One valid and one invalid order reference
    cursor.executemany(
        "INSERT INTO order_items VALUES (?, ?, ?)",
        [
            ("ORD001", "P001", 2),
            ("ORD999", "P002", 1)
        ]
    )

    # Find invalid references
    cursor.execute("""
        SELECT oi.order_id
        FROM order_items oi
        LEFT JOIN orders o
            ON oi.order_id = o.order_id
        WHERE o.order_id IS NULL
    """)

    invalid_orders = cursor.fetchall()

    conn.close()

    assert len(invalid_orders) == 1
    assert invalid_orders[0][0] == "ORD999"

    print("PASS: Invalid order_id was detected.")
```

```python
test_invalid_order_reference()
```

```python
def test_discount_greater_than_100():
    discount_percent = 150

    assert discount_percent <= 100, \
        "Invalid discount: discount_percent cannot be greater than 100"

    print("PASS: Invalid discount was handled.")
```

```python
def test_discount_greater_than_100():
    discount_percent = 150

    if discount_percent > 100:
        result = "INVALID"
    else:
        result = "VALID"

    assert result == "INVALID"

    print("PASS: Discount greater than 100 was rejected.")
```

```python
#Test 3 — Quantity = 0
def test_zero_quantity():
    quantity = 0
    unit_price = 1000
    discount_percent = 10

    revenue = (
        quantity
        * unit_price
        * (1 - discount_percent / 100)
    )

    assert revenue == 0

    print("PASS: Zero quantity results in zero revenue.")
```

```python
#Test 4 — Future order date
def test_future_order_date():
    future_date = "2099-01-01 10:30:00"

    order_datetime = datetime.strptime(
        future_date,
        "%Y-%m-%d %H:%M:%S"
    )

    current_datetime = datetime.now()

    assert order_datetime > current_datetime

    print("PASS: Future order date was detected.")
```
